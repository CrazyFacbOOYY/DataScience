# Задача 7
Написать функцию, принимающая целое число n, задающее количество кубиков. Функция
должна определить, можно ли из данного кол-ва кубиков построить пирамиду, то есть
можно ли представить число n как 1^2+2^2+3^2+…+k^2. Если можно, то функция должна
вернуть k, иначе строку “It is impossible”. Написать тесты для кода

```python
def pyramid(number):
    if type(number) not in [int]:
        print("It is impossible")

    if number <= 0:
        print("It is impossible")

    k = 1
    while True:
        total = k * (k + 1) * (2 * k + 1) // 6
        
        if total == number:
            return k
        elif total > number:
            return "It is impossible"
        k += 1

def test_succesfull():
    successful_values = [1,5,14,30,55]

    for index, value in enumerate(successful_values):
        assert pyramid(value) == index
    
def test_failed(self):
    failed_values = [2,3,6,8,10,12,23,35,60]
    for index, value in enumerate(failed_values):
        assert pyramid(value) == "It is impossible"

def test_incorrectTypes(self):
    incorrect_inputs = ["string", 1.1]
    for index, value in enumerate(incorrect_inputs):
        assert pyramid(value) == "It is impossible"

    def test_negateOrZeroValues(self):
        incorrect_inputs = [-1, 0]
        for index, value in enumerate(incorrect_inputs):
            assert pyramid(value) == "It is impossible"
```

# Задача 8
Функция принимает на вход положительное число и определяет является ли оно сбалансированным, т.е. сумма цифр до средних равна сумме цифр после. Средними в случае нечетного числа цифр считать одну цифру, в случае четного - две средних. Написать тесты для кода

```python
def is_balanced_number(number):
    stroke_Digit = str(number)
    length = len(stroke_Digit)

    left_sum = 0
    right_sum = 0
    
    if length <= 2:
        return True
    
    if length % 2 == 0:
        mid_start = length // 2 - 1
        mid_end = length // 2 + 1
    else:
        mid_start = length // 2
        mid_end = length // 2 + 1

    left_part = stroke_Digit[:mid_start]
    right_part = stroke_Digit[mid_end:]
    
    for digit in left_part:
        left_sum += int(digit)
    for digit in right_part:
        right_sum += int(digit)
    
    return left_sum == right_sum

def test_single_digit():
    assert is_balanced_number(1)
    assert is_balanced_number(9)
    assert is_balanced_number(0)
    
def test_two_digits(self):
    assert is_balanced_number(12)
    assert is_balanced_number(99)
    assert is_balanced_number(10)
    
def test_balanced(self):
    assert is_balanced_number(121)
    assert is_balanced_number(12321)
    assert is_balanced_number(10203)
    
def test_unbalanced(self):
    assert not is_balanced_number(123)
    assert not is_balanced_number(12345)
    assert not is_balanced_number(10234)
    
def test_length_balanced(self):
    assert is_balanced_number(1221)
    assert is_balanced_number(123321)
    assert is_balanced_number(1001)
    
def test_length_unbalanced(self):
    assert not is_balanced_number(1234)
    assert not is_balanced_number(123456)
    assert not is_balanced_number(1002)
```
